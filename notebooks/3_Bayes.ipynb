## Exercise: The Rare Disease Test 🩺

This exercise will help you understand how new evidence changes a belief using a real-world scenario.

---

### The Scenario

Imagine a very rare disease called "Digitalitis." It's known that only **1 in every 1,000 people** has it.

A new medical test is developed to detect Digitalitis. The test is pretty good, but not perfect:
* It's **99% accurate** at correctly identifying people who *have* the disease (this is the "true positive" rate).
* It has a **5% "false positive" rate**. This means 5% of perfectly healthy people will incorrectly test positive.

You take the test and your result is **positive**.

**Question:** Before doing any math, what does your gut tell you? What's the chance you actually have Digitalitis? Write it down!

---

### The Calculation: A Town of 1,000 People

To figure this out, let's imagine we administer this test to a random sample of **1,000 people**.



Use the information from the scenario to answer the following questions.

1.  **The Sick vs. The Healthy:**
    * Out of 1,000 people, how many would you expect to actually have Digitalitis?
    * How many people are healthy?

2.  **Calculating the Positive Tests:**
    * Let's look at the sick group first. How many of the sick people will test positive? (Hint: The test is 99% accurate for them). This is the number of **true positives**.
    * Now, let's look at the healthy group. How many of the healthy people will test positive? (Hint: The false positive rate is 5%). This is the number of **false positives**.

3.  **The Final Result:**
    * What is the total number of people who tested positive (true positives + false positives)?
    * Out of all the people who tested positive, how many *actually have the disease*?
    * So, what is the real probability ($\frac{\text{People who are actually sick}}{\text{Total people who tested positive}}$) that you have Digitalitis given your positive test?

---

### Conclusion

Compare the final probability to the number you wrote down at the beginning. Are you surprised?

This is the core of Bayes' formula: it correctly combines an **initial belief** (the disease is very rare) with **new evidence** (the positive test result) to give a more accurate, updated belief.

## Testing a Group for Digitalitis

Let's test a big group of **1,000 people** to see what happens.

---

### Step 1: How many are sick?  
- **Disease rate:** 1 in 1,000  
- So, **1 person** actually has Digitalitis.  
- The other **999 people are healthy**.  

---

### Step 2: How many will test positive?

- The **1 sick person** will almost certainly test positive  
  - Test accuracy for sick people: **99%**  
  - ✅ **1 true positive**  

- The **999 healthy people** will also get tested  
  - False positive rate: **5%**  
  - 999 × 0.05 ≈ **50 false positives**  

---

### Step 3: Total test positives
- **1 true positive**  
- **50 false positives**  
- **Total: 51 people** tested positive  

---

### Step 4: Who actually has the disease?

Out of **51 positive results**:  
- **Only 1 person** really has the disease.  

So the probability is:  

\begin{eqnarray*}
P(\text{Disease} \mid \text{Positive Test}) = \frac{1}{51} \approx 2\%
\end{eqnarray*}

---

### Step 5: Understanding the change in belief
- **Before the test (prior probability):** 0.1%  
- **After the test (posterior probability):** ~2%  

Your belief increased, but it’s still nowhere near 99%.  
This is because the disease is **very rare**, and Bayes’ theorem incorporates that rarity.

---

### Bayes’ Formula

\begin{eqnarray*}
P(\text{Disease} \mid \text{Positive Test}) = 
\frac{P(\text{Positive Test} \mid \text{Disease}) \cdot P(\text{Disease})}
{P(\text{Positive Test})}
\end{eqnarray*}

Where:  
- **$P(\text{Disease} \mid \text{Positive Test})$:** Updated belief (~2%)  
- **$P(\text{Disease})$:** Initial belief (0.1% rarity)  
- **$P(\text{Positive Test} \mid \text{Disease})$:** Test’s accuracy (99%)  


## Exercise

During the Vietnam War in 1975, researchers needed to evaluate the proportion of soldiers who smoke. Since smoking was a sensitive topic that soldiers might not answer honestly, they designed an experiment to protect individual privacy. When asked "Do you smoke?", each soldier was instructed to first toss a coin privately. Only the soldier could see the result. If the coin showed heads, the soldier was instructed to answer "yes" regardless of whether they actually smoke or not. If the coin showed tails, the soldier was instructed to answer truthfully.

After 200 responses, 160 soldiers answered "yes" and 40 answered "no".

Can you compute the probability that a soldier is a smoker (using Bayes' theorem) ?

What is the probability that you smoke given that you said yes?

## Question 1: What proportion of soldiers smoke?

**Given:**
- 160 out of 200 said "yes" → P(yes) = 0.8

**Logic:**
A soldier says "yes" if:
- Coin is heads (always say yes) → probability 0.5
- OR coin is tails AND they smoke → probability 0.5 × P(smoker)

**Equation:**
P(yes) = 0.5 + 0.5 × P(smoker)

0.8 = 0.5 + 0.5P(smoker)

0.3 = 0.5P(smoker)

**P(smoker) = 0.6**

### Answer: **60% of soldiers are smokers**

---

## Question 2: P(smoker | said "yes")?

**Using Bayes' Theorem:**

P(smoker | yes) = P(yes | smoker) × P(smoker) / P(yes)

**Calculate P(yes | smoker):**
If you're a smoker:
- Heads → say yes
- Tails → say yes (truthful)
- P(yes | smoker) = 1.0

**Plug in:**
P(smoker | yes) = 1.0 × 0.6 / 0.8 = 0.75

### Answer: **75% chance that someone who said "yes" is actually a smoker**

---

## Intuition

- Only 80% said yes, but the coin forces 50% to say yes
- The "extra" 30% are the true smokers with tails
- So 30% = half of the smokers, meaning 60% total smoke
- Of those who said yes, 3/4 are real smokers

Below I asked Claude to answer the question with simulations.

```python
import random

# Observed data: 160 "yes" out of 200 soldiers
observed_yes = 160
total_soldiers = 200
observed_yes_rate = observed_yes / total_soldiers

print("=== FINDING THE TRUE SMOKING RATE ===\n")

# Try different smoking rates and see which one produces results closest to observed
best_smoking_rate = 0
best_difference = float('inf')

for test_rate in range(0, 101):
    smoking_rate = test_rate / 100
    
    # Simulate many times with this smoking rate
    simulations = 10000
    total_yes = 0
    
    for _ in range(simulations):
        yes_count = 0
        
        # Simulate 200 soldiers
        for _ in range(total_soldiers):
            # Is this soldier a smoker?
            is_smoker = random.random() < smoking_rate
            
            # Flip coin
            coin = random.choice(['heads', 'tails'])
            
            # Determine answer
            if coin == 'heads':
                answer = 'yes'  # Always say yes if heads
            else:  # tails
                answer = 'yes' if is_smoker else 'no'  # Tell truth
            
            if answer == 'yes':
                yes_count += 1
        
        total_yes += yes_count
    
    # Average yes rate from simulations
    simulated_yes_rate = total_yes / (simulations * total_soldiers)
    difference = abs(simulated_yes_rate - observed_yes_rate)
    
    if difference < best_difference:
        best_difference = difference
        best_smoking_rate = smoking_rate

print(f"Observed 'yes' rate: {observed_yes_rate:.1%}")
print(f"Best matching smoking rate: {best_smoking_rate:.1%}")
print(f"This means approximately {best_smoking_rate:.1%} of soldiers are smokers\n")

print("=== VERIFYING WITH SIMULATION ===\n")

# Run a large simulation with the best rate to verify
simulations = 100000
yes_count = 0
for _ in range(simulations):
    is_smoker = random.random() < best_smoking_rate
    coin = random.choice(['heads', 'tails'])
    
    if coin == 'heads':
        answer = 'yes'
    else:
        answer = 'yes' if is_smoker else 'no'
    
    if answer == 'yes':
        yes_count += 1

print(f"With {best_smoking_rate:.1%} smoking rate:")
print(f"Simulated 'yes' rate: {yes_count/simulations:.1%}")
print(f"Expected: {observed_yes_rate:.1%}\n")

print("=== P(SMOKER | SAID YES) ===\n")

# Simulate and track both smoking status and answer
simulations = 100000
said_yes_count = 0
said_yes_and_smoker_count = 0

for _ in range(simulations):
    is_smoker = random.random() < best_smoking_rate
    coin = random.choice(['heads', 'tails'])
    
    if coin == 'heads':
        answer = 'yes'
    else:
        answer = 'yes' if is_smoker else 'no'
    
    if answer == 'yes':
        said_yes_count += 1
        if is_smoker:
            said_yes_and_smoker_count += 1

prob_smoker_given_yes = said_yes_and_smoker_count / said_yes_count

print(f"Out of {said_yes_count:,} who said 'yes':")
print(f"{said_yes_and_smoker_count:,} were actual smokers")
print(f"\nP(Smoker | Said Yes) = {prob_smoker_given_yes:.1%}")
print(f"\nThis means if someone said 'yes', there's a {prob_smoker_given_yes:.1%}")
print(f"chance they're actually a smoker (not just forced to say yes by the coin)")
```

Here is the code from Claude when asked to be more efficient and **the code is much better !**

```python
import numpy as np

def simulate_randomized_response(smoking_rate, n_sims):
    """
    Simulate the randomized response protocol.
    Returns: (is_smoker, answer_yes) boolean arrays
    """
    is_smoker = np.random.random(n_sims) < smoking_rate
    coin_heads = np.random.random(n_sims) < 0.5
    answer_yes = coin_heads | is_smoker
    return is_smoker, answer_yes

# Observed data: 160 "yes" out of 200 soldiers
observed_yes = 160
total_soldiers = 200
observed_yes_rate = observed_yes / total_soldiers

print("=== FINDING THE TRUE SMOKING RATE ===\n")

# Try different smoking rates and see which one produces results closest to observed
test_rates = np.linspace(0, 1, 101)
best_smoking_rate = 0
best_difference = float('inf')

for smoking_rate in test_rates:
    _, answer_yes = simulate_randomized_response(smoking_rate, n_sims=1000000)
    
    simulated_yes_rate = answer_yes.mean()
    difference = abs(simulated_yes_rate - observed_yes_rate)
    
    if difference < best_difference:
        best_difference = difference
        best_smoking_rate = smoking_rate

print(f"Observed 'yes' rate: {observed_yes_rate:.1%}")
print(f"Best matching smoking rate: {best_smoking_rate:.1%}")
print(f"This means approximately {best_smoking_rate:.1%} of soldiers are smokers\n")

print("=== VERIFYING WITH SIMULATION ===\n")

# Verify with large vectorized simulation
_, answer_yes = simulate_randomized_response(best_smoking_rate, n_sims=1000000)

print(f"With {best_smoking_rate:.1%} smoking rate:")
print(f"Simulated 'yes' rate: {answer_yes.mean():.1%}")
print(f"Expected: {observed_yes_rate:.1%}\n")

print("=== P(SMOKER | SAID YES) ===\n")

# Calculate P(Smoker | Said Yes) using vectorized operations
is_smoker, answer_yes = simulate_randomized_response(best_smoking_rate, n_sims=1000000)

# Count those who said yes and are smokers
said_yes_and_smoker = is_smoker & answer_yes
prob_smoker_given_yes = said_yes_and_smoker.sum() / answer_yes.sum()

print(f"Out of {answer_yes.sum():,} who said 'yes':")
print(f"{said_yes_and_smoker.sum():,} were actual smokers")
print(f"\nP(Smoker | Said Yes) = {prob_smoker_given_yes:.1%}")
print(f"\nThis means if someone said 'yes', there's a {prob_smoker_given_yes:.1%}")
print(f"chance they're actually a smoker (not just forced to say yes by the coin)")
```
