## Exercise: The Rare Disease Test 🩺

This exercise will help you understand how new evidence changes a belief using a real-world scenario.

---

### The Scenario

Imagine a very rare disease called "Digitalitis." It's known that only **1 in every 1,000 people** has it.

A new medical test is developed to detect Digitalitis. The test is pretty good, but not perfect:
* It's **99% accurate** at correctly identifying people who *have* the disease (this is the "true positive" rate).
* It has a **5% "false positive" rate**. This means 5% of perfectly healthy people will incorrectly test positive.

You take the test and your result is **positive**.

**Question:** Before doing any math, what does your gut tell you? What's the chance you actually have Digitalitis? Write it down!

---

### The Calculation: A Town of 1,000 People

To figure this out, let's imagine we administer this test to a random sample of **1,000 people**.



Use the information from the scenario to answer the following questions.

1.  **The Sick vs. The Healthy:**
    * Out of 1,000 people, how many would you expect to actually have Digitalitis?
    * How many people are healthy?

2.  **Calculating the Positive Tests:**
    * Let's look at the sick group first. How many of the sick people will test positive? (Hint: The test is 99% accurate for them). This is the number of **true positives**.
    * Now, let's look at the healthy group. How many of the healthy people will test positive? (Hint: The false positive rate is 5%). This is the number of **false positives**.

3.  **The Final Result:**
    * What is the total number of people who tested positive (true positives + false positives)?
    * Out of all the people who tested positive, how many *actually have the disease*?
    * So, what is the real probability ($\frac{\text{People who are actually sick}}{\text{Total people who tested positive}}$) that you have Digitalitis given your positive test?

---

### Conclusion

Compare the final probability to the number you wrote down at the beginning. Are you surprised?

This is the core of Bayes' formula: it correctly combines an **initial belief** (the disease is very rare) with **new evidence** (the positive test result) to give a more accurate, updated belief.
